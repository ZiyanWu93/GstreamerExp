# H13 — Does a shorter keyframe interval soften the loss cliff?

**Status:** `untested` · **Source:** Goal 2 boundary behaviour (§7.8), validates the loss-collapse diagnostic (H10/H12)


## Claim

The loss-collapse diagnostic concluded that H10/H12's delivery cliff at 1% loss is a receiver-side VP8 reference-chain desync gated by the 60-frame keyframe interval — one early gap breaks inter-frame prediction and the stream cannot re-anchor before the next loss re-breaks it. The falsifiable prediction is that shortening the keyframe interval bounds the desync window, so complete-frame delivery should rise as the GOP shrinks. Holding H10's 1% recovery-off cell fixed and sweeping the keyframe interval over 60, 30, 15, 8, 4, 2 frames, delivery should climb well above the ~18% long-GOP baseline. If delivery stays collapsed regardless of GOP, the GOP is not the cause and the diagnostic is wrong.

## Predictions

- `shorter_gop_raises_delivery`

## Findings and Limitations

**Limitations**

- Pre-registered: predictions and the verdict rule are fixed before the run. Results pending the sweep on the aum/veda testbed.
- Single variable. Only encoder.keyframe_interval_frames changes across the six cells; loss (1% uniform), network, ceiling, recovery (off), and workload are held at H10's cell, so any delivery change is attributable to the GOP.
- Trade-off, not a free lunch. Shorter GOPs spend more of the 4000 kbps budget on keyframes, so PSNR-of-delivered may fall even as delivery rises; the boundary signal is delivery, and PSNR is reported alongside.
- Confounded by frame size. A shorter GOP makes each frame larger (more packets), raising per-frame loss probability, which works against delivery; if delivery still rises, the re-anchoring effect dominates the larger-frame effect.
- Workload-specific (realmotion-avi, 1280x1024 MJPEG, 10 fps, 60 s), recovery off. 3 reps per cell. Boundary signal is frame delivery.
