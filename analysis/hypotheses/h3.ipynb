# H3 — The test video is too lightweight to compare SCReAM and GCC on 5G recordings

**Status:** `supported` · **Source:** Project-internal validity check for 5G-recording SCReAM / GCC experiments


## Claim

The current test video (ball-720p30) sends nowhere near enough data to use up the bandwidth in the 5G recordings, so any controller comparison on those recordings is not yet meaningful.

## Predictions

- `every_5g_recording_has_at_least_10x_more_capacity_than_the_camera_actually_sends`
- `on_every_recording_fewer_than_15_percent_of_100ms_windows_drop_below_1mbps_of_actual_sent_bytes`

## Verdict

| Outcome | Predicate |

## Findings and Limitations

**Findings**

- The current ball-720p30 test video sends about 220–230 kbit / s, well below the typical bandwidth in the translated 5G recordings.
- Earlier 5G-recording SCReAM / GCC comparisons cannot be read as meaningful algorithm comparisons.
- Any future 5G-recording comparison must first pass this calibration with a test video that actually keeps the link busy.

**Limitations**

- This shows the current test video is too light. It does not yet test a heavier replacement.
- Brief low-capacity dips still happen in the handover-like recordings; the claim is about typical bandwidth across the whole replay.
