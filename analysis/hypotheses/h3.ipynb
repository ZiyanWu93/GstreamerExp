# H3 — Media workload calibration gates realistic 5G trace comparisons

**Status:** `supported` · **Source:** Project-internal validity check for realistic-trace SCReAM/GCC experiments


## Claim

Media workload calibration gates realistic 5G trace comparisons.

## Prediction

The current ball-720p30-30s VP8 workload is not valid for comparing SCReAM and GCC on the translated 5G traces because observed camera wire bitrate is far below trace capacity. A workload is considered under-driving when each translated 5G trace has at least 10x median capacity headroom over observed camera wire bitrate and fewer than 15% of 100 ms bins below 1 Mbit/s.


## Verdict rule

Supported if all tested realistic traces meet the under-driving rule; inconclusive if only some do; refuted if none do; untested if required records are absent.


## Findings and Limitations

**Findings**

- The current ball-720p30-30s VP8 workload emits roughly 220-230 kbit/s, far below the typical capacity of the translated 5G traces.
- The previous realistic-trace SCReAM/GCC comparisons should not be interpreted as meaningful algorithm comparisons.
- A future realistic-trace comparison must first pass this calibration gate with a high-quality or high-bitrate workload whose observed wire bitrate actually stresses the trace.

**Limitations**

- This validates that the old workload is too light; it does not yet test a replacement high-quality workload.
- Brief low-capacity fades still exist in the handover-like traces; the claim is about typical headroom across the full replay.
